# Day 4 - Data Cleaning and Validation

```python
# Check duplicate records

duplicate_count = df.duplicated().sum()

print("Total duplicate records:", duplicate_count)

# Display duplicate records
df[df.duplicated(keep=False)].head(10)
```

```python
# Count occurrences of each exact record

duplicate_groups = df[
    df.duplicated(keep=False)
].value_counts().reset_index(name="Occurrences")

print("Number of duplicate groups:", len(duplicate_groups))

print("\nMost repeated records:")
display(duplicate_groups.head(10))
```

```python
# Create a cleaned copy

df_clean = df.drop_duplicates().copy()

print("Original dataset shape:", df.shape)
print("Cleaned dataset shape:", df_clean.shape)

print("Duplicates remaining:", df_clean.duplicated().sum())
```

```python
# Check missing values

missing_values = df_clean.isnull().sum()

missing_report = pd.DataFrame({
    "Missing Values": missing_values,
    "Percentage": (missing_values / len(df_clean)) * 100
})

print(missing_report)
```

```python
# Display rows with missing descriptions

missing_description = df_clean[
    df_clean["Description"].isnull()
]

display(missing_description.head(10))

print("Missing descriptions:", len(missing_description))
```

```python
# Display rows with missing Customer IDs

missing_customer = df_clean[
    df_clean["Customer ID"].isnull()
]

display(missing_customer.head(10))

print("Missing Customer IDs:", len(missing_customer))
```

```python
# Remove rows with missing descriptions

df_clean = df_clean.dropna(subset=["Description"]).copy()

print("Dataset shape after handling missing descriptions:", df_clean.shape)

print("\nRemaining missing values:")
print(df_clean.isnull().sum())
```

```python
# Check Customer ID values

print("Missing Customer IDs:", df_clean["Customer ID"].isnull().sum())

print("Unique identified customers:",
      df_clean["Customer ID"].nunique())
```

```python
# Identify cancelled invoices

cancelled = df_clean[
    df_clean["Invoice"].astype(str).str.startswith("C")
]

print("Cancelled transaction records:", len(cancelled))

print("\nCancelled quantity total:",
      cancelled["Quantity"].sum())

print("\nSample cancelled transactions:")
display(cancelled.head(10))
```

```python
# Calculate cancelled invoice count

cancelled_invoice_count = cancelled["Invoice"].nunique()

normal_invoice_count = df_clean[
    ~df_clean["Invoice"].astype(str).str.startswith("C")
]["Invoice"].nunique()

print("Cancelled invoices:", cancelled_invoice_count)
print("Normal invoices:", normal_invoice_count)
```

```python
# Separate normal sales and cancellation records

df_sales = df_clean[
    ~df_clean["Invoice"].astype(str).str.startswith("C")
].copy()

df_returns = df_clean[
    df_clean["Invoice"].astype(str).str.startswith("C")
].copy()

print("Normal sales shape:", df_sales.shape)
print("Cancellation records shape:", df_returns.shape)

print("Sales records remaining:",
      df_sales["Invoice"].nunique())

print("Cancellation invoices:",
      df_returns["Invoice"].nunique())
```

```python
# Investigate zero-price records

zero_price = df_clean[df_clean["Price"] == 0]

print("Zero-price records:", len(zero_price))

display(zero_price.head(10))
```

```python
# Investigate negative-price records

negative_price = df_clean[df_clean["Price"] < 0]

print("Negative-price records:", len(negative_price))

display(negative_price)
```

```python
# Filter ordinary positive sales

df_sales_clean = df_sales[
    (df_sales["Quantity"] > 0) &
    (df_sales["Price"] > 0)
].copy()

# Calculate revenue

df_sales_clean["Revenue"] = (
    df_sales_clean["Quantity"] * df_sales_clean["Price"]
)

print("Clean sales dataset shape:", df_sales_clean.shape)

print("Remaining zero prices:",
      (df_sales_clean["Price"] == 0).sum())

print("Remaining negative prices:",
      (df_sales_clean["Price"] < 0).sum())

print("Remaining negative quantities:",
      (df_sales_clean["Quantity"] < 0).sum())

print("Total sales revenue:",
      round(df_sales_clean["Revenue"].sum(), 2))
```

```python
# Final data validation

print("Dataset shape:", df_sales_clean.shape)

print("\nMissing values:")
print(df_sales_clean.isnull().sum())

print("\nDuplicate records:",
      df_sales_clean.duplicated().sum())

print("\nData types:")
print(df_sales_clean.dtypes)

print("\nDate range:")
print(df_sales_clean["InvoiceDate"].min(),
      "to",
      df_sales_clean["InvoiceDate"].max())

print("\nRevenue summary:")
print(df_sales_clean["Revenue"].describe())
```

```python
# Save cleaned sales dataset

output_path = "data/processed/cleaned_sales.csv"

df_sales_clean.to_csv(output_path, index=False)

print("Cleaned dataset saved successfully!")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

df_sales_clean = pd.read_csv(
    "data/processed/cleaned_sales.csv",
    parse_dates=["InvoiceDate"],
    dtype={
        "Invoice": str,
        "StockCode": str
    }
)

print("Dataset loaded successfully!")
print("Shape:", df_sales_clean.shape)
```

```python
# Visualize numerical distributions
plt.figure(figsize=(15, 5))

for i, column in enumerate(
    ["Quantity", "Price", "Revenue"], 1
):
    plt.subplot(1, 3, i)
    sns.histplot(
        df_sales_clean[column],
        bins=50,
        kde=True
    )
    plt.title(f"{column} Distribution")
    plt.xlabel(column)

plt.tight_layout()
plt.show()
```

```python
for column in ["Quantity", "Price", "Revenue"]:

    Q1 = df_sales_clean[column].quantile(0.25)
    Q3 = df_sales_clean[column].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df_sales_clean[
        (df_sales_clean[column] < lower) |
        (df_sales_clean[column] > upper)
    ]

    print(f"\n{column}")
    print("Lower Bound:", round(lower, 2))
    print("Upper Bound:", round(upper, 2))
    print("Outlier Count:", len(outliers))
    print("Outlier Percentage:",
          round(len(outliers) / len(df_sales_clean) * 100, 2), "%")
```

```python
# Calculate RFM metrics

rfm = rfm_data.groupby("Customer ID").agg(
    Recency=("InvoiceDate", 
             lambda x: (reference_date - x.max()).days),

    Frequency=("Invoice", "nunique"),

    Monetary=("Revenue", "sum")
).reset_index()

# Display first 10 customers
rfm.head(10)
```

```python
# Assign RFM scores using quintiles

rfm["R_Score"] = pd.qcut(
    rfm["Recency"].rank(method="first"),
    5,
    labels=[5, 4, 3, 2, 1]
)

rfm["F_Score"] = pd.qcut(
    rfm["Frequency"].rank(method="first"),
    5,
    labels=[1, 2, 3, 4, 5]
)

rfm["M_Score"] = pd.qcut(
    rfm["Monetary"].rank(method="first"),
    5,
    labels=[1, 2, 3, 4, 5]
)

# Convert scores to integers
rfm[["R_Score", "F_Score", "M_Score"]] = (
    rfm[["R_Score", "F_Score", "M_Score"]]
    .astype(int)
)

rfm.head(10)
```

```python
rfm["RFM_Score"] = (
    rfm["R_Score"].astype(str) +
    rfm["F_Score"].astype(str) +
    rfm["M_Score"].astype(str)
)

rfm["RFM_Total"] = (
    rfm["R_Score"] +
    rfm["F_Score"] +
    rfm["M_Score"]
)

rfm.head(10)
```

```python
def segment_customer(row):
    if row["RFM_Score"] == "555":
        return "Best Customers"
    elif row["R_Score"] >= 4 and row["F_Score"] >= 4:
        return "Loyal Customers"
    elif row["R_Score"] >= 4 and row["F_Score"] <= 2:
        return "Recent Customers"
    elif row["R_Score"] <= 2 and row["F_Score"] >= 4:
        return "At Risk Customers"
    elif row["R_Score"] <= 2 and row["F_Score"] <= 2:
        return "Lost Customers"
    elif row["M_Score"] >= 4:
        return "Big Spenders"
    else:
        return "Regular Customers"

rfm["Customer_Segment"] = rfm.apply(segment_customer, axis=1)

rfm[[
    "Customer ID",
    "RFM_Score",
    "RFM_Total",
    "Customer_Segment"
]].head(20)
```

# Day 23 - Customer Retention Strategy

## Objective
Identify customers at risk of churn and develop targeted
customer retention strategies using ML predictions and SHAP.

```python
import pandas as pd
import numpy as np

# Generate churn probabilities
churn_probabilities = best_rf_model.predict_proba(X_test)[:, 1]

# Create customer prediction results
customer_predictions = pd.DataFrame({
    "Customer_ID": model_data.loc[X_test.index, "Customer ID"].values,
    "Actual_Churn": y_test.values,
    "Churn_Probability": churn_probabilities,
    "Predicted_Churn": best_rf_model.predict(X_test)
})

# Display results
display(customer_predictions.head(10))
```

```python
def classify_risk(probability):
    if probability < 0.30:
        return "Low Risk"
    elif probability <= 0.60:
        return "Medium Risk"
    else:
        return "High Risk"

customer_predictions["Risk_Category"] = (
    customer_predictions["Churn_Probability"].apply(classify_risk)
)

display(
    customer_predictions[
        ["Customer_ID", "Churn_Probability",
         "Predicted_Churn", "Risk_Category"]
    ].head(15)
)
```

```python
risk_summary = customer_predictions["Risk_Category"].value_counts()

display(risk_summary)
```

## Customer Retention Strategies

### 1. High Risk
- Personalized retention offers
- Special discounts
- Direct customer follow-up
- Win-back campaigns

### 2. Medium Risk
- Product recommendations
- Email engagement campaigns
- Loyalty reward reminders
- Limited-time offers

### 3. Low Risk
- Loyalty programmes
- Cross-selling recommendations
- Referral rewards
- Exclusive customer benefits

```python
retention_strategies = {
    "High Risk": "Win-back campaigns and personalized discounts",
    "Medium Risk": "Engagement campaigns and product recommendations",
    "Low Risk": "Loyalty rewards and referral programmes"
}

customer_predictions["Retention_Strategy"] = (
    customer_predictions["Risk_Category"].map(retention_strategies)
)

display(customer_predictions.head(15))
```

```python
risk_evaluation = customer_predictions.groupby(
    "Risk_Category"
).agg(
    Total_Customers=("Customer_ID", "count"),
    Average_Churn_Probability=("Churn_Probability", "mean"),
    Actual_Churned=("Actual_Churn", "sum"),
    Actual_Churn_Rate=("Actual_Churn", "mean")
).reset_index()

risk_evaluation["Average_Churn_Probability"] *= 100
risk_evaluation["Actual_Churn_Rate"] *= 100

display(risk_evaluation.round(2))
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=customer_predictions,
    x="Risk_Category",
    order=["Low Risk", "Medium Risk", "High Risk"]
)

plt.title("Customer Distribution by Churn Risk")
plt.xlabel("Risk Category")
plt.ylabel("Number of Customers")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9, 5))

sns.barplot(
    data=risk_evaluation,
    x="Risk_Category",
    y="Actual_Churn_Rate"
)

plt.title("Actual Churn Rate by Risk Category")
plt.xlabel("Risk Category")
plt.ylabel("Actual Churn Rate (%)")
plt.tight_layout()
plt.show()
```
